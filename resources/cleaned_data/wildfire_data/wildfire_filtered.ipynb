```python
#import dependencies and libraries
import pandas as pd
import numpy as np

#read in the data
fires_2020=pd.read_csv('crime_2020_wildfire.csv')
fires_2021=pd.read_csv('crime_2021_wildfire.csv')
fires_2022=pd.read_csv('crime_2022_wildfire.csv')
fires_2023=pd.read_csv('crime_2023_wildfire.csv')
fires_2024=pd.read_csv('crime_2024_wildfire.csv')

```

```python
#build a loop to iterate through the dataframes and isolate the the values during the wildfire event
filtered_wildfires_list = []
for df in [fires_2020, fires_2021, fires_2022, fires_2023, fires_2024]:
    #filter rows where the wildfire event is occuring
    filtered_wildfires = df[df['Wildfire Event'] == 1]
    filtered_wildfires_list.append(filtered_wildfires)

#concatenate the dataframes
wildfire_crime = pd.concat(filtered_wildfires_list, ignore_index=True)

#export the data to a csv
wildfire_crime.to_csv('wildfire_crime_filtered.csv', index=False)
```
